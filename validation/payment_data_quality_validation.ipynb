```python
%sql
SELECT COUNT(*) AS csv_records
FROM payment_catalog.payment_de.bronze_payments;
```

```python
%sql
SELECT COUNT(*) AS json_records
FROM payment_catalog.payment_de.bronze_payments_json;
```

```python
%sql
SELECT
    COUNT(*) AS total_records,
    COUNT_IF(rejection_reason IS NULL) AS valid_records,
    COUNT_IF(rejection_reason IS NOT NULL) AS invalid_records
FROM payment_catalog.payment_de.silver_payments_validated;
```

```python
%sql
SELECT rejection_reason, COUNT(*) AS rejected_records
FROM payment_catalog.payment_de.silver_payments_rejected
GROUP BY rejection_reason;
```

```python
%sql
SELECT
    COUNT(*) AS total_records,
    COUNT(DISTINCT transaction_id) AS unique_transactions
FROM payment_catalog.payment_de.silver_payments_deduplicated;
```

```python
%sql
SELECT *
FROM payment_catalog.payment_de.gold_payment_kpis;
```

```python
%sql
SELECT
    COUNT(*) AS total_transactions,
    ROUND(SUM(amount), 2) AS total_amount,
    COUNT_IF(is_fraud = 1) AS fraud_count,
    ROUND(
        100.0 * COUNT_IF(is_fraud = 1) / NULLIF(COUNT(*), 0),
        2
    ) AS fraud_rate
FROM payment_catalog.payment_de.silver_payments_deduplicated;
```

```python
%sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT transaction_id) AS unique_ids,
    COUNT_IF(transaction_id IS NULL) AS null_ids
FROM payment_catalog.payment_de.silver_payments_deduplicated;
```
